**Transcrição**

```
from groq import Groq
import json

cliente = Groq(api_key=GROQ_API_KEY)

arquivo = "audio.wav"

with open(arquivo, "rb") as audio:
    transcricao = cliente.audio.transcriptions.create(
        file = audio,
        model = "whisper-large-v3-turbo",
        timestamp_granularities = ["segment"],
        response_format="verbose_json"
    )

print(transcricao.text)
```

```
from IPython.display import display, Markdown
import json

# Montar a saída em Markdown
saida = f"""
# 🎙️ Transcrição do áudio
---

## 📝 Texto reconhecido

> {transcricao.text}
"""

# Exibir os segmentos com marcação de tempo, se disponíveis
if getattr(transcricao, "segments", None):
    saida += "\n\n## ⏱️ Segmentos do áudio\n\n"
    for segmento in transcricao.segments:
        inicio = segmento["start"] if isinstance(segmento, dict) else segmento.start
        fim = segmento["end"] if isinstance(segmento, dict) else segmento.end
        texto = segmento["text"] if isinstance(segmento, dict) else segmento.text

        saida += f"- **{inicio:.2f}s – {fim:.2f}s:** {texto.strip()}\n"

display(Markdown(saida))
```

```
import requests

# coleção de frases e traduções
url = "https://tatoeba.org/pt-br/audio/download/1274624"

resposta = requests.get(url)
resposta.raise_for_status()

with open("frase.mp3", "wb") as arquivo:
    arquivo.write(resposta.content)

print("Download concluído: frase.mp3")
```

```
from groq import Groq

cliente = Groq(api_key=GROQ_API_KEY)

arquivo = "frase.mp3"

with open(arquivo, "rb") as audio:
    traducao = cliente.audio.translations.create(
        file = (arquivo, audio.read()),
        model = "whisper-large-v3"
    )

print(traducao.text)
```

### **Visão**

> https://console.groq.com/docs/vision

```
import requests
import base64
from groq import Groq

cliente = Groq(api_key = GROQ_API_KEY)

url = "https://www.fatosdesconhecidos.com.br/wp-content/uploads/2023/10/selecao.webp"

resposta_imagem = requests.get(url, headers={"User-Agent": "Mozilla/5.0"})
resposta_imagem.raise_for_status()

imagem_base64 = base64.b64encode(resposta_imagem.content).decode("utf-8")

resposta = cliente.chat.completions.create(
    model="qwen/qwen3.8-27b",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "text",
                    "text": "Descreva em português o que aparece nesta imagem."
                },
                {
                    "type": "image_url",
                    "image_url": {
                        "url": f"data:image/jpeg;base64,{imagem_base64}"
                    }
                }
            ]
        }
    ],
    max_completion_tokens=1024
)
```

```
print(resposta.choices[0].message.content)
```

```
from IPython.display import display, Markdown

display(Markdown(resposta.choices[0].message.content))
```

### **Saídas Estruturadas**

> https://console.groq.com/docs/structured-outputs

```
from groq import Groq
import json

cliente = Groq(api_key=GROQ_API_KEY)

resposta = cliente.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "system",
            "content": "Extraia informações sobre o produto a partir do texto de uma avaliação."
        },
        {
            "role": "user",
            "content": (
                "Comprei os fones de ouvido UltraSound na semana passada "
                "e estou muito impressionado! O cancelamento de ruído é "
                "excelente e a bateria dura o dia inteiro. A qualidade do "
                "som é nítida e clara. Eu daria 4,5 de 5 estrelas."
            )
        }
    ],
    response_format={
        "type": "json_schema",
        "json_schema": {
            "name": "avaliacao_produto",
            "strict": True,
            "schema": {
                "type": "object",
                "properties": {
                    "nome_produto": {
                        "type": "string"
                    },
                    "avaliacao": {
                        "type": "number"
                    },
                    "sentimento": {
                        "type": "string",
                        "enum": ["positivo", "negativo", "neutro"]
                    },
                    "caracteristicas_principais": {
                        "type": "array",
                        "items": {
                            "type": "string"
                        }
                    }
                },
                "required": [
                    "nome_produto",
                    "avaliacao",
                    "sentimento",
                    "caracteristicas_principais"
                ],
                "additionalProperties": False
            }
        }
    }
)
```

```
resultado = json.loads(resposta.choices[0].message.content or "{}")

print(json.dumps(resultado, indent=2, ensure_ascii=False))
```

**Classificação de E-mail**

```
from groq import Groq
from pydantic import BaseModel
import json

cliente = Groq(api_key = GROQ_API_KEY)

class EntidadePrincipal(BaseModel):
    entidade: str
    tipo: str

class ClassificacaoEmail(BaseModel):
    categoria: str
    prioridade: str
    pontuacao_confianca: float
    sentimento: str
    entidades_principais: list[EntidadePrincipal]
    acoes_sugeridas: list[str]
    requer_atencao_imediata: bool
    tempo_estimado_resposta: str

resposta = cliente.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {
            "role": "system",
            "content": (
                "Você é especialista em classificação de e-mails. "
                "Classifique os e-mails em categorias estruturadas, "
                "incluindo pontuação de confiança, níveis de prioridade "
                "e ações sugeridas."
            )
        },
        {
            "role": "user",
            "content": (
                "Assunto: URGENTE: Indisponibilidade do servidor afetando a produção\n\n"
                "Olá, equipe,\n\n"
                "Nosso principal servidor de produção ficou indisponível "
                "às 14h30, no horário da Costa Leste dos Estados Unidos. "
                "Os serviços disponíveis aos clientes estão atualmente "
                "fora do ar. Precisamos agir imediatamente para restaurar "
                "os serviços. Por favor, participem da chamada de emergência.\n\n"
                "Atenciosamente,\n"
                "Equipe de DevOps"
            )
        }
    ],
    response_format={
        "type": "json_schema",
        "json_schema": {
            "name": "classificacao_email",
            "schema": ClassificacaoEmail.model_json_schema()
        }
    }
)
```

```
classificacao_email = ClassificacaoEmail.model_validate_json(
    resposta.choices[0].message.content
)

print(json.dumps(
    classificacao_email.model_dump(),
    indent=2,
    ensure_ascii=False
))
```

**Análise de Sentimento**

```
from groq import Groq
import json

cliente = Groq(api_key=GROQ_API_KEY)

resposta = cliente.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {
            "role": "system",
            "content": """Você é uma API de análise de dados que realiza análise de sentimentos em textos.
Responda somente em JSON, utilizando o seguinte formato:
{
    "analise_sentimento": {
        "sentimento": "positivo|negativo|neutro",
        "pontuacao_confianca": 0.95,
        "expressoes_principais": [
            {
                "expressao": "expressão principal identificada",
                "sentimento": "positivo|negativo|neutro"
            }
        ],
        "resumo": "Uma frase que resume o sentimento geral"
    }
}"""
        },
        {
            "role": "user",
            "content": (
                "Analise o sentimento desta avaliação de cliente: "
                "'Eu simplesmente adoro este produto! A qualidade "
                "superou minhas expectativas, embora a entrega tenha "
                "demorado mais do que o esperado.'"
            )
        }
    ],
    response_format={"type": "json_object"}
)
```

```
resultado = json.loads(resposta.choices[0].message.content)

print(json.dumps(resultado, indent=2, ensure_ascii=False))
```
