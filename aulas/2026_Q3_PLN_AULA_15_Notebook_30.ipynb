## **Similaridade de Textos**
---



**Similaridade de Textos** refere-se à medida de quão semelhantes são dois textos. Essa medida pode ser usada para comparar diferentes documentos, avaliar a originalidade de um texto ou detectar plágio.

Há muitas aplicações práticas para a similaridade de textos, incluindo:

   * Detecção de plágio: a similaridade de textos pode ser usada para detectar plágio em trabalhos acadêmicos, teses, artigos ou outros tipos de documentos. A comparação da similaridade entre um documento e uma base de dados de textos pode indicar se há evidência de plágio.

   * Recomendação de conteúdo: a similaridade de textos pode ser usada para recomendar conteúdo a usuários com base em suas preferências. Por exemplo, se um usuário gosta de um determinado tipo de artigo ou livro, a similaridade de texto pode ser usada para recomendar conteúdo semelhante.

   * Busca de informações: a similaridade de textos pode ser usada para ajudar na busca de informações. Por exemplo, se um usuário procura informações sobre um determinado tópico, a similaridade de texto pode ser usada para encontrar documentos semelhantes que possam fornecer informações relevantes.

   * Análise de sentimentos: a similaridade de textos pode ser usada para analisar sentimentos em comentários ou avaliações de usuários. A comparação da similaridade entre as opiniões de diferentes usuários pode indicar tendências ou opiniões gerais sobre um produto ou serviço.

   * Classificação de documentos: a similaridade de textos pode ser usada para classificar documentos em diferentes categorias com base em seu conteúdo. Por exemplo, se uma empresa tem um grande volume de documentos, a similaridade de texto pode ser usada para classificar esses documentos de acordo com suas categorias ou temas.

   * Identificação de fraudes: a similaridade de textos pode ser usada para identificar fraudes ou atividades suspeitas em documentos financeiros ou transações bancárias. A comparação da similaridade entre documentos pode ajudar a identificar possíveis irregularidades ou padrões suspeitos.

   * Análise de concorrência: a similaridade de textos pode ser usada para analisar a concorrência em um mercado ou setor específico. A comparação da similaridade entre os sites, publicidades e materiais promocionais de diferentes empresas pode ajudar a identificar tendências e estratégias de marketing.

   * Resumos automáticos: a similaridade de textos pode ser usada para gerar resumos automáticos de documentos longos. A comparação da similaridade entre diferentes seções de um documento pode ajudar a identificar as partes mais relevantes e importantes para incluir no resumo.

### **Métricas de Similaridade de Textos**



**Similaridade de Hamming**

A **distância de Hamming** é uma métrica que mede a diferença entre duas sequências de igual comprimento, contando o número de posições em que essas sequências diferem.

Para calcular a distância de Hamming entre duas sequências, é necessário comparar cada posição correspondente nas duas sequências e contar o número de posições em que os valores são diferentes.

A distância de Hamming é uma métrica muito simples e útil para detectar erros em transmissões de dados, onde pode ser usada para detectar a presença de bits errados ou perdidos.

```
def distancia_hamming(palavra1, palavra2):
    """Calcula a distância de Hamming entre duas palavras."""
    if len(palavra1) != len(palavra2):
        raise ValueError("As palavras precisam ter o mesmo comprimento.")
    return sum(c1 != c2 for c1, c2 in zip(palavra1, palavra2))
```

```
# Exemplo de uso

palavra1 = "gato"
palavra2 = "gata"

distancia = distancia_hamming(palavra1, palavra2)
print(f"A distância de Hamming entre '{palavra1}' e '{palavra2}' é {distancia}.")
```

Neste exemplo, a função `distancia_hamming` recebe duas palavras `'palavra1'` e `'palavra2'` e calcula a distância de Hamming entre elas. A função compara cada letra nas duas palavras e conta o número de posições em que elas diferem. Em seguida, a distância de Hamming é retornada. O código também inclui um exemplo de uso da função, onde a distância de Hamming entre as palavras `"gato"` e `"gata"` é calculada e impressa.

A distância de Hamming é uma medida discreta que representa o número de posições em que as duas sequências são diferentes. No entanto, é possível calcular uma `"distância de Hamming normalizada"` dividindo a distância de Hamming pelo comprimento das sequências, o que resulta em um valor entre 0 e 1 que indica a proporção de posições em que as duas sequências são diferentes. Esse valor pode ser interpretado como uma medida de dissimilaridade ou `distância relativa` entre as duas sequências.

```
def hamming_distance_normalized(s1, s2):
    """Calcula a distância de Hamming normalizada entre duas sequências."""
    if len(s1) != len(s2):
        raise ValueError("As sequências precisam ter o mesmo comprimento.")
    d = sum(c1 != c2 for c1, c2 in zip(s1, s2))
    return d / len(s1)
```

```
# Exemplo de uso

s1 = "abcd"
s2 = "abed"

distancia = hamming_distance_normalized(s1, s2)
print(f"A distância de Hamming normalizada entre '{s1}' e '{s2}' é {distancia:.2f}.")
```

**Similaridade de Jaccard**

A **Similaridade de Jaccard** é uma medida de similaridade entre dois conjuntos que avalia a proporção de elementos que são compartilhados entre eles. A similaridade de Jaccard é definida como a interseção do conjunto dividida pela união dos conjuntos.

A fórmula da similaridade de Jaccard é a seguinte:

> `J(A, B) = |A ∩ B| / |A ∪ B|`

onde A e B são dois conjuntos, A ∩ B é a interseção de A e B (ou seja, o conjunto de elementos que estão em ambos os conjuntos) e A ∪ B é a união de A e B (ou seja, o conjunto de elementos que estão em pelo menos um dos conjuntos).

O resultado da similaridade de Jaccard é um valor entre 0 e 1, em que 0 significa que não há elementos compartilhados entre os conjuntos e 1 significa que os conjuntos são idênticos. Quanto maior a similaridade de Jaccard, mais semelhantes são os conjuntos.

A similaridade de Jaccard é comumente usada em áreas como mineração de dados, aprendizado de máquina e PLN para comparar a similaridade entre conjuntos de itens.

Aqui está um exemplo simples de como calcular a similaridade de Jaccard entre dois conjuntos:

```
def jaccard_similarity(set1, set2):
    """
    Calcula a similaridade de Jaccard entre dois conjuntos.
    """
    intersection = set1.intersection(set2)
    union = set1.union(set2)
    return len(intersection) / len(union)
```

```
# Exemplo de uso

conjunto_frutas1 = {'maçã', 'banana', 'laranja', 'morango'}
conjunto_frutas2 = {'maçã', 'uva', 'kiwi'}

similaridade = jaccard_similarity(conjunto_frutas1, conjunto_frutas2)

print(f"Similaridade de Jaccard entre os conjuntos: {similaridade:.2f}")
```

Aqui está um exemplo de como usar a similaridade de Jaccard para comparar duas frases:

```
def jaccard_similarity(text1, text2):
    """
    Calcula a similaridade de Jaccard entre duas frases.
    """
    # Transforma as frases em conjuntos de palavras únicas
    set1 = set(text1.split())
    set2 = set(text2.split())

    # Calcula a interseção e a união dos conjuntos de palavras
    intersection = set1.intersection(set2)
    union = set1.union(set2)

    # Retorna a similaridade de Jaccard
    return len(intersection) / len(union)
```

```
# Exemplo de uso

frase1 = "O gato está dormindo."
frase2 = "O cachorro está acordado."

similaridade = jaccard_similarity(frase1, frase2)

print(f"Similaridade de Jaccard entre as frases: {similaridade:.2f}")
```

**Distância Euclidiana**

A **distância euclidiana** é uma medida de distância entre dois pontos em um espaço n-dimensional, que pode ser usada para calcular a distância entre dois vetores de características que representam dois textos. A distância euclidiana é calculada como a raiz quadrada da soma dos quadrados das diferenças entre as coordenadas dos vetores.

Para aplicar a distância euclidiana à similaridade de textos, primeiro é necessário representar cada texto como um vetor de características. Existem várias maneiras de fazer isso, como representar cada palavra como uma dimensão do vetor e a contagem de ocorrências de cada palavra como o valor da coordenada, ou usar técnicas de embedding para representar cada palavra como um vetor contínuo de dimensão fixa.

Suponha que temos dois textos representados como vetores de características `x` e `y`, cada um com `n` dimensões. A distância euclidiana entre os vetores `x` e `y` é dada pela seguinte fórmula:

> `d(x, y) = sqrt((x1 - y1)^2 + (x2 - y2)^2 + ... + (xn - yn)^2)`

Onde `x1, x2, ..., xn` são as coordenadas do vetor `x` e `y1, y2, ..., yn` são as coordenadas do vetor `y`. A raiz quadrada é usada para obter uma medida de distância positiva.



Para calcular a similaridade de textos usando a distância euclidiana, podemos usar a fórmula inversa da distância, que é dada por:

> `s(x, y) = 1 / (1 + d(x, y))`

onde `s(x, y)` é a similaridade entre os textos x e y. A fórmula acima retorna um valor entre 0 e 1, onde 0 indica que os textos são totalmente diferentes e 1 indica que os textos são idênticos. O uso da fórmula inversa da distância garante que quanto menor a distância euclidiana entre os vetores, maior será a similaridade entre os textos.

Segue abaixo um exemplo de código para calcular a similaridade de dois textos utilizando a distância euclidiana:

```
import math

def distancia_euclidiana(vetor1, vetor2):
    # Calcula a distância euclidiana entre os dois vetores
    distancia = math.sqrt(sum([(vetor1[i]-vetor2[i])**2 for i in range(len(vetor1))]))
    return distancia

def similaridade_euclidiana(texto1, texto2):
    # Calcula a similaridade entre os dois textos utilizando a distância euclidiana
    palavras = list(set(texto1.split() + texto2.split()))
    vetor1 = [texto1.split().count(palavra) for palavra in palavras]
    vetor2 = [texto2.split().count(palavra) for palavra in palavras]
    distancia = distancia_euclidiana(vetor1, vetor2)
    similaridade = 1 / (1 + distancia)
    return similaridade
```

```
# Exemplo de uso

texto1 = "O gato comeu o rato."
texto2 = "O cachorro comeu a ração."

similaridade = similaridade_euclidiana(texto1, texto2)

print(f"Similaridade euclidiana entre os textos: {similaridade:.4f}")
```

**Similaridade de Cosseno**

A **similaridade de cosseno** é uma medida de similaridade entre dois vetores, muito utilizada na área de PLN para comparar a similaridade entre documentos.

A ideia por trás da similaridade de cosseno é comparar o ângulo formado pelos vetores no espaço de n dimensões onde n é o número de termos distintos presentes nos documentos. A similaridade é calculada a partir do cosseno do ângulo entre esses vetores. Se o ângulo entre dois vetores for pequeno, então os vetores são considerados semelhantes e a similaridade de cosseno será próxima de 1. Se o ângulo for próximo a 90 graus, então os vetores são considerados dissimilares e a similaridade de cosseno será próxima de 0.



Em resumo, a similaridade de cosseno mede a semelhança entre dois vetores através do cosseno do ângulo formado entre eles. Quando aplicada a documentos, cada documento é representado como um vetor de termos, e a similaridade de cosseno é utilizada para comparar a similaridade entre esses vetores.

`TF-IDF` é uma medida de peso que atribui maior importância aos termos que aparecem com frequência em um documento, mas que aparecem raramente em outros documentos. `TF-IDF` é calculado multiplicando a frequência de um termo no documento (TF) pelo inverso da frequência desse termo em todo o conjunto de documentos (IDF).

A similaridade entre dois textos é então calculada utilizando a pontuação `TF-IDF` dos termos que aparecem em ambos os textos. Existem várias formas de calcular a similaridade usando `TF-IDF`, como a similaridade de cosseno, que mede a similaridade entre dois vetores de pesos de termos `TF-IDF`.

Aqui está um exemplo de código para calcular a similaridade de cosseno entre dois textos usando a biblioteca `scikit-learn` e a técnica `TF-IDF`:

```
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Textos de exemplo
texto1 = "O Brasil é um país localizado na América do Sul."
texto2 = "O Japão é um país localizado na Ásia."

# Transformação dos textos em vetores utilizando TF-IDF
vetorizador = TfidfVectorizer()
vetores_tfidf = vetorizador.fit_transform([texto1, texto2])

# Cálculo da similaridade de cosseno entre os vetores
similaridade = cosine_similarity(vetores_tfidf[0], vetores_tfidf[1])

# Exibição do resultado
print(f"Similaridade de cosseno entre os textos: {similaridade[0][0]:.4f}")
```

Uma outra biblioteca que pode ser usada para cálculo de similaridade de cosseno é a biblioteca `numpy`. Aqui está um exemplo de código usando `numpy` para calcular a similaridade de cosseno entre dois vetores de texto:

```
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

# Textos de exemplo
texto1 = "O Brasil é um país localizado na América do Sul."
texto2 = "O Japão é um país localizado na Ásia."

# Transformação dos textos em vetores utilizando TF-IDF
vetorizador = TfidfVectorizer()
vetores_tfidf = vetorizador.fit_transform([texto1, texto2]).toarray()

# Cálculo da similaridade de cosseno
produto_escalar = np.dot(vetores_tfidf[0], vetores_tfidf[1])

norma_texto1 = np.linalg.norm(vetores_tfidf[0])
norma_texto2 = np.linalg.norm(vetores_tfidf[1])

similaridade = produto_escalar / (norma_texto1 * norma_texto2)

# Exibição do resultado
print(f"Similaridade de cosseno entre os textos: {similaridade:.4f}")
```

Para criar uma função que recomende títulos de filmes baseado na descrição semelhante usando similaridade de textos, podemos utilizar `TF-IDF` para representar cada descrição em um vetor numérico, e depois calcular a similaridade entre as descrições usando a distância do cosseno.

Para isso, podemos seguir os seguintes passos:

   1. Importar as bibliotecas necessárias: `pandas` para manipulação de dataframes, `scikit-learn` para o cálculo do `TF-IDF` e da similaridade de cosseno, e `numpy` para operações numéricas.

   2. Criar uma função que recebe como entrada um título de filme e um dataframe contendo títulos e descrições de filmes, e retorna uma lista com os títulos de filmes recomendados com base na similaridade de descrições.

   3. Na função, usar o título de entrada para selecionar a descrição correspondente no dataframe.

   4. Usar o `TF-IDFVectorizer` da biblioteca `scikit-learn` para calcular a representação `TF-IDF` de todas as descrições no dataframe.

   5. Calcular a matriz de similaridade de cosseno entre as descrições usando a função `cosine_similarity` da biblioteca `scikit-learn`.

   6. Selecionar as linhas da matriz de similaridade correspondentes à descrição do título de entrada e ordená-las em ordem decrescente de similaridade.

   7. Selecionar os títulos correspondentes às linhas selecionadas e adicioná-los a uma lista de recomendações.

   8. Retornar a lista de recomendações.

Segue abaixo uma possível implementação dessa função:

```
import nltk

nltk.download('stopwords')
```

```
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def recomendar_filmes(titulo, df, n=5):
    # seleciona a descrição correspondente ao título de entrada
    descricao = df[df['titulo'] == titulo]['descricao'].values[0]

    stopwords = nltk.corpus.stopwords.words('portuguese')

    # calcula a representação TF-IDF de todas as descrições, desconsiderando as stopwords
    vectorizer = TfidfVectorizer(stop_words=stopwords)
    tfidf_matrix = vectorizer.fit_transform(df['descricao'])

    # calcula a matriz de similaridade de cosseno entre as descrições
    sim_matrix = cosine_similarity(tfidf_matrix)

    # seleciona as linhas correspondentes à descrição do título de entrada
    idx = df[df['titulo'] == titulo].index[0]
    sim_scores = list(enumerate(sim_matrix[idx]))

    # ordena as linhas em ordem decrescente de similaridade
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # seleciona os títulos correspondentes às linhas selecionadas e adiciona a uma lista de recomendações
    recomendacoes = []
    for i in range(1, n+1):
        title = df.iloc[sim_scores[i][0]]['titulo']
        sim = sim_scores[i][1]
        recomendacoes.append((title, sim))

    return recomendacoes
```

Essa função recebe três argumentos: o título de entrada, um dataframe contendo títulos e descrições de filmes, e o número de filmes a serem recomendados (padrão é 5).

```
recomendacoes = recomendar_filmes("O Poderoso Chefão", df, n=5)

# Exibe os títulos dos filmes recomendados e seus respectivos valores de similaridade
print("Filmes recomendados:\n")

for titulo, similaridade in recomendacoes:
    print(f"Filme: {titulo} | Similaridade: {similaridade:.4f}")
```

**Similaridade semântica usando *embeddings***

Um *embedding* representa um texto como um vetor numérico. Textos com significados próximos tendem a produzir vetores próximos no espaço vetorial.

A similaridade de cosseno compara a orientação de dois vetores. Seu valor varia de -1 a 1; em *embeddings* de texto, valores mais próximos de 1 geralmente indicam maior proximidade semântica.

```
MODELO_EMBEDDING = "text-embedding-3-small"
```

```
import numpy as np
from openai import OpenAI

cliente = OpenAI( api_key= OPENAI_API_KEY )

def gerar_embedding(texto, modelo=MODELO_EMBEDDING):
    resposta = cliente.embeddings.create(
        model = modelo,
        input = texto
    )
    return np.array(resposta.data[0].embedding, dtype=float)

def similaridade_cosseno(vetor_a, vetor_b):
    denominador = np.linalg.norm(vetor_a) * np.linalg.norm(vetor_b)
    if denominador == 0:
        return 0.0
    return float(np.dot(vetor_a, vetor_b) / denominador)
```

```
# Exemplo de uso

texto1 = "O aluno está estudando inteligência artificial."
texto2 = "O estudante está aprendendo sobre IA."
texto3 = "A previsão do tempo indica chuva amanhã."

emb1 = gerar_embedding(texto1)
emb2 = gerar_embedding(texto2)
emb3 = gerar_embedding(texto3)

print(f"Texto 1 x Texto 2: {similaridade_cosseno(emb1, emb2):.4f}")
print(f"Texto 1 x Texto 3: {similaridade_cosseno(emb1, emb3):.4f}")
```

Altere os textos e observe como a similaridade se modifica. Compare, por exemplo, frases com palavras diferentes, mas com significado próximo, e frases com vocabulário parecido, mas sentidos diferentes.

```
frases = [
    "O cachorro correu pelo parque.",
    "Um cão estava correndo em uma área verde.",
    "O gato dormiu no sofá.",
    "A bolsa de valores registrou alta."
]

vetores = [gerar_embedding(frase) for frase in frases]

for i in range(len(frases)):
    for j in range(i + 1, len(frases)):
        score = similaridade_cosseno(vetores[i], vetores[j])
        print(f"{score:.4f} | {frases[i]}  <->  {frases[j]}")
```

O próximo exemplo recebe uma consulta e compara seu *embedding* com os *embeddings* de uma coleção de textos. Os resultados são ordenados pela similaridade de cosseno.

```
documentos = [
    "A inteligência artificial pode apoiar o diagnóstico médico.",
    "Redes neurais são utilizadas em tarefas de reconhecimento de imagens.",
    "A computação em nuvem oferece recursos computacionais sob demanda.",
    "O processamento de linguagem natural permite analisar textos.",
    "Algoritmos de aprendizado de máquina identificam padrões em dados."
]

embeddings_documentos = [gerar_embedding(doc) for doc in documentos]
```

```
def buscar_similares(consulta, textos, embeddings, top_k=3):
    vetor_consulta = gerar_embedding(consulta)
    resultados = []

    for texto, vetor in zip(textos, embeddings):
        score = similaridade_cosseno(vetor_consulta, vetor)
        resultados.append((texto, score))

    return sorted(resultados, key=lambda item: item[1], reverse=True)[:top_k]
```

```
consulta = "Como computadores conseguem compreender a linguagem humana?"

for texto, score in buscar_similares(
    consulta, documentos, embeddings_documentos, top_k=3
):
    print(f"{score:.4f} | {texto}")
```

Nesse caso, é recomendável dividir o conteúdo em trechos (*chunks*), gerar *embeddings* por trecho e definir uma estratégia de agregação ou recuperação.

**Similaridade com um Grande Modelo de Linguagem**

Outra possibilidade é solicitar ao modelo de linguagem que avalie a relação entre dois textos. Diferentemente da similaridade de cosseno, a resposta pode incluir uma classificação e uma justificativa.

A avaliação abaixo utiliza uma escala de 0 a 5 e critérios explícitos. A nota gerada pelo modelo é uma avaliação interpretativa, não uma medida matemática equivalente à similaridade de cosseno.

```
clinte = OpenAI( api_key= OPENAI_API_KEY )

def avaliar_com_llm(texto_a, texto_b):
    prompt = f"""
Compare os dois textos considerando o significado, e não apenas as palavras.

Atribua uma nota de 0 a 5:
0 = sem relação temática ou semântica
1 = relação muito fraca
2 = relação parcial
3 = relação moderada
4 = significados bastante próximos
5 = significados equivalentes

Explique brevemente a avaliação. Diferencie textos que tratam do
mesmo assunto daqueles que expressam a mesma informação.

Texto A:
{texto_a}

Texto B:
{texto_b}

Apresente a resposta em português, com nota e justificativa.
"""

    resposta = cliente.responses.create(
        model = "gpt-6-luna",
        input = prompt
    )

    return resposta.output_text
```

**Comparar abordagens: *embeddings* e avaliação por LLM**

As duas abordagens respondem a perguntas relacionadas, mas não idênticas:

- **Embeddings + cosseno:** fornecem uma medida numérica eficiente, útil para ranqueamento, busca semântica e comparação em coleções maiores.

A escolha depende do objetivo da aplicação. Em sistemas de busca, por exemplo, *embeddings* são frequentemente usados para recuperar candidatos; um modelo de linguagem pode ser aplicado posteriormente para reordenar ou justificar os resultados.

**Criar um Banco Vetorial**

Um banco vetorial (*Vector Store*) é uma estrutura utilizada para armazenar e consultar documentos representados por vetores numéricos, chamados *embeddings*.

Esses vetores representam características semânticas dos textos, permitindo encontrar documentos com significados semelhantes, mesmo quando utilizam palavras diferentes.

Como funciona?
1. Armazenamento: os textos são convertidos em *embeddings* por um modelo de linguagem.
2. Indexação: os vetores são armazenados no banco vetorial.
3. Consulta: uma pergunta ou texto de busca também é convertido em *embedding*.
4. Busca: o banco compara o vetor da consulta com os vetores armazenados e recupera os documentos mais semelhantes.

```
documentos = [
    Document(
        page_content="A inteligência artificial pode apoiar o diagnóstico médico.",
        metadata={"tema": "saúde"}
    ),
    Document(
        page_content="Redes neurais são utilizadas no reconhecimento de imagens.",
        metadata={"tema": "IA"}
    ),
    Document(
        page_content="A computação em nuvem oferece recursos sob demanda.",
        metadata={"tema": "computação"}
    ),
    Document(
        page_content="O processamento de linguagem natural permite analisar textos.",
        metadata={"tema": "PLN"}
    ),
    Document(
        page_content="Algoritmos de aprendizado de máquina identificam padrões em dados.",
        metadata={"tema": "aprendizado de máquina"}
    )
]

banco_vetorial = InMemoryVectorStore(modelo_embeddings)

banco_vetorial.add_documents( documents=documentos )
```

**Busca Semântica**

A busca semântica é uma técnica de recuperação de informações que procura identificar textos relacionados ao significado de uma consulta, e não apenas aqueles que contêm as mesmas palavras.

Diferentemente de uma busca tradicional por palavras-chave, a busca semântica consegue recuperar documentos que utilizam palavras diferentes, mas apresentam significados semelhantes.

```
consulta = "Como computadores podem interpretar a linguagem humana?"

resultados = banco_vetorial.similarity_search( consulta, k=2 )

for i, documento in enumerate(resultados, start=1):
    print(f"{i}. {documento.page_content}")
    print("Metadados:", documento.metadata)
```

**Retriever**

No **LangChain**, um `Retriever` é um componente responsável por recuperar documentos relevantes a partir de uma consulta (pergunta ou texto).

Ele utiliza uma estratégia de recuperação, como similaridade semântica, para localizar documentos relacionados à consulta em uma coleção de dados.

```
retriever = banco_vetorial.as_retriever( search_kwargs={"k": 2} )

documentos_recuperados = retriever.invoke(
    "Quais são as aplicações de IA em textos?"
)

for documento in documentos_recuperados:
    print(documento.page_content)
```

**Busca semântica com RAG**

A **busca semântica com RAG** (*Retrieval-Augmented Generation*) combina a recuperação de informações por similaridade de significado com a geração de respostas por um modelo de linguagem.

A ideia é permitir que um modelo responda a perguntas utilizando informações recuperadas de uma coleção de documentos, em vez de depender exclusivamente do conhecimento adquirido durante seu treinamento.

```
prompt_rag = ChatPromptTemplate.from_template(
    """
Responda à pergunta usando exclusivamente o contexto fornecido.
Se o contexto não contiver informação suficiente, informe isso.

Contexto:
{contexto}

Pergunta:
{pergunta}

Resposta:
"""
)

def formatar_documentos(documentos):
    return "\n\n".join(
        doc.page_content for doc in documentos
    )

def recuperar_contexto(entrada):
    docs = retriever.invoke(entrada["pergunta"])

    return {
        "pergunta": entrada["pergunta"],
        "contexto": formatar_documentos(docs)
    }

cadeia_rag = (
    recuperar_contexto
    | prompt_rag
    | modelo
    | StrOutputParser()
)
```

```
resposta = cadeia_rag.invoke({
    "pergunta": "Como a inteligência artificial pode ser utilizada na análise de textos?"
})

print(resposta)
```

Um exemplo completo:

```
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_core.documents import Document
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough, RunnableLambda

# 1. Modelo de embeddings
embeddings = OpenAIEmbeddings( model="text-embedding-3-small", api_key = OPENAI_API_KEY )

# 2. Documentos
documentos = [
    Document(page_content="A inteligência artificial permite automatizar tarefas e processos."),
    Document(page_content="O aprendizado de máquina identifica padrões em conjuntos de dados."),
    Document(page_content="O processamento de linguagem natural permite analisar textos e compreender informações linguísticas."),
    Document(page_content="As redes neurais artificiais são utilizadas em tarefas de classificação e previsão.")
]

# 3. Criar e alimentar o banco vetorial
banco = InMemoryVectorStore(embeddings)
banco.add_documents(documentos)

# 4. Criar o retriever
retriever = banco.as_retriever(
    search_kwargs={"k": 2}
)

# 5. Modelo de linguagem
modelo = ChatOpenAI( model="gpt-6-luna", api_key = OPENAI_API_KEY )

# 6. Prompt
prompt_rag = ChatPromptTemplate.from_template(
    """
Responda à pergunta usando exclusivamente o contexto fornecido.
Se o contexto não contiver informação suficiente, informe isso.

Contexto:
{contexto}

Pergunta:
{pergunta}

Resposta:
"""
)

# 7. Construir a cadeia RAG
cadeia_rag = (
    {
        "contexto": retriever | RunnableLambda(
            lambda docs: "\n\n".join(
                doc.page_content for doc in docs
            )
        ),
        "pergunta": RunnablePassthrough()
    }
    | prompt_rag
    | modelo
    | StrOutputParser()
)

# 8. Realizar uma consulta
pergunta = "Como a inteligência artificial pode ser utilizada na análise de textos?"

resposta = cadeia_rag.invoke(pergunta)

print(resposta)
```

**Em resumo**

A busca semântica é uma das técnicas utilizadas na etapa de recuperação do RAG. O RAG amplia esse processo ao fornecer os documentos recuperados como contexto para um modelo de linguagem, que gera uma resposta fundamentada nessas informações.

**Similaridade com `ChatPromptTemplate`**

```
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

modelo = ChatOpenAI( model="gpt-6-luna", api_key = OPENAI_API_KEY )

prompt_comparacao = ChatPromptTemplate.from_messages([
    (
        "system",
        """
Você avalia similaridade semântica entre textos.

Considere o significado, não apenas as palavras.

Utilize a seguinte escala:

0 - nenhuma relação
1 - relação muito fraca
2 - relação parcial
3 - similaridade moderada
4 - alta similaridade
5 - significados praticamente equivalentes

Explique a nota e diferencie assunto semelhante
de informação equivalente.

Responda em português.
"""
    ),
    (
        "human",
        """
Texto A:
{texto_a}

Texto B:
{texto_b}

Apresente a nota e a justificativa.
"""
    )
])

cadeia_comparacao = (
    prompt_comparacao
    | modelo
    | StrOutputParser()
)
```

```
resultado = cadeia_comparacao.invoke({
    "texto_a": "O aluno utiliza um chatbot para estudar programação.",
    "texto_b": "Um estudante aprende programação com o auxílio de um assistente conversacional."
})
```

### **Exemplo: Sistema de Perguntas e Respostas sobre uma Biblioteca Universitária**

**Importação das bibliotecas**



```
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_core.documents import Document
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough, RunnableLambda

from IPython.display import display, Markdown
```

**Criação dos documentos**

Cada documento representa uma informação sobre o funcionamento da biblioteca.

```
documentos = [
    Document(
        page_content="Os alunos de graduação podem retirar até cinco livros simultaneamente, pelo prazo de 14 dias.",
        metadata={"categoria": "emprestimo"}
    ),
    Document(
        page_content="Os livros podem ser renovados por mais 14 dias, desde que não existam reservas realizadas por outros usuários.",
        metadata={"categoria": "renovacao"}
    ),
    Document(
        page_content="Livros devolvidos com atraso estão sujeitos a uma suspensão temporária do direito de realizar novos empréstimos.",
        metadata={"categoria": "atraso"}
    ),
    Document(
        page_content="A biblioteca funciona de segunda a sexta-feira, das 8h às 22h, e aos sábados, das 8h às 13h.",
        metadata={"categoria": "horario"}
    ),
    Document(
        page_content="Para realizar empréstimos, o estudante deve apresentar sua identificação institucional.",
        metadata={"categoria": "emprestimo"}
    ),
    Document(
        page_content="Os computadores da biblioteca podem ser utilizados pelos estudantes para pesquisas acadêmicas e atividades de estudo.",
        metadata={"categoria": "infraestrutura"}
    ),
    Document(
        page_content="As salas de estudo em grupo devem ser reservadas previamente pelo sistema da biblioteca.",
        metadata={"categoria": "infraestrutura"}
    )
]
```

**Criação do banco vetorial**

Os documentos são convertidos em *embeddings* e armazenados no banco vetorial.

```
embeddings = OpenAIEmbeddings( model="text-embedding-3-small", api_key=OPENAI_API_KEY )

banco = InMemoryVectorStore(embeddings)

banco.add_documents(documentos)
```

**Configuração do `Retriever`**

O `Retriever` será responsável por recuperar os três documentos mais relevantes para cada pergunta.

```
retriever = banco.as_retriever(
    search_kwargs={"k": 3}
)
```

**Configuração do modelo e do prompt**

O prompt estabelece que as respostas devem utilizar exclusivamente as informações recuperadas.

```
modelo = ChatOpenAI( model="gpt-6-luna", api_key = OPENAI_API_KEY )

prompt_rag = ChatPromptTemplate.from_template(
    """
Você é um assistente virtual de uma biblioteca universitária.

Responda à pergunta utilizando exclusivamente as informações
apresentadas no contexto.

Não invente informações.
Se o contexto não apresentar informações suficientes,
informe que não foi possível localizar essa informação.

Contexto:
{contexto}

Pergunta:
{pergunta}

Resposta:
"""
)
```

**Construção da cadeia RAG**

A cadeia realiza a recuperação dos documentos, organiza o contexto, combina as informações com a pergunta e envia o resultado ao modelo de linguagem.

```
cadeia_rag = (
    {
        "contexto": retriever | RunnableLambda(
            # recebe os documentos recuperados e reúne seus conteúdos em uma única string
            lambda docs: "\n\n".join(
                doc.page_content for doc in docs
            )
        ),
        # simplesmente repassa a entrada original, sem modificá-la
        "pergunta": RunnablePassthrough()
    }
    | prompt_rag
    | modelo
    | StrOutputParser()
)
```

**Realização de uma consulta**

```
pergunta = "Quantos livros posso retirar e por quanto tempo?"

resposta = cadeia_rag.invoke(pergunta)

display(Markdown(f"""
### Pergunta

{pergunta}

### Resposta

{resposta}
"""))
```

**Visualização dos documentos recuperados**

Também é possível verificar quais documentos foram recuperados pelo `Retriever` para responder à pergunta.

```
documentos_recuperados = retriever.invoke(pergunta)

display(Markdown("### Documentos recuperados"))

for i, doc in enumerate(documentos_recuperados, start=1):
    display(Markdown(f"""
**Documento {i}**

{doc.page_content}

<small>Categoria: {doc.metadata['categoria']}</small>
"""))
```

**Testando diferentes perguntas**

Uma mesma cadeia pode responder a diferentes perguntas sobre os documentos armazenados.

```
perguntas = [
    "Qual é o horário de funcionamento da biblioteca?",
    "Posso ficar com um livro por mais de duas semanas?",
    "O que acontece se eu devolver um livro atrasado?",
    "Como posso utilizar as salas de estudo em grupo?",
    "A biblioteca oferece algum recurso para utilizar computadores?"
]

for i, pergunta in enumerate(perguntas, start=1):

    resposta = cadeia_rag.invoke(pergunta)

    display(Markdown(f"""
### Pergunta {i}

{pergunta}

**Resposta:**

{resposta}

---
"""))
```
